```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("../results/grounded_molecules.tsv", sep="\t")

df = df[df["MOL_SCORE"] != "Not Available"]
dfchebi = df[df["MOL_TYPE"] == "CHEBI"]
scores = pd.to_numeric(dfchebi["MOL_SCORE"])
```

```python
plt.hist(scores, bins=40, color="steelblue", edgecolor="white")

# Adding a vertical line at the mean score value
plt.axvline(
    scores.mean(), color="red", linestyle="--", label=f"Mean = {scores.mean():.3f}"
)

plt.title("Distribution of Scores for Chebi")
plt.xlabel("Score")
plt.ylabel("Number of entries")
plt.legend()
```
