```python
# Notebook analyzes the score distribution from the grounding results.
# in `ground_mol_chebi.tsv`.

import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("../results/ground_mol_chebi.tsv", sep="\t")
df2 = pd.read_csv("../results/ground_mol_gilda.tsv", sep="\t")

# Replace all "not found" strings with actual NaN, then drop them
not_found_values = ["NOT_FOUND", "Not Found", "NA", "NULL", "NaN"]

df = df.replace(not_found_values, pd.NA)
df2 = df2.replace(not_found_values, pd.NA)

scores = pd.to_numeric(df["Score"], errors="coerce").dropna()
scores2 = pd.to_numeric(df2["Score"], errors="coerce").dropna()
```

```python
# Plot the score distribution as a histogram using 40 bins (from the Chebi results)
plt.hist(scores, bins=40, color="steelblue", edgecolor="white")

# Adding a vertical line at the mean score value
plt.axvline(
    scores.mean(), color="red", linestyle="--", label=f"Mean = {scores.mean():.3f}"
)

# Set the title and labels for the plot
plt.title("Distribution of Scores for Chebi")
plt.xlabel("Score")
plt.ylabel("Number of entries")
plt.legend()
```

```python
# Plot the score distribution as a histogram using 40 bins (from the Gilda results)
plt.hist(scores2, bins=40, color="darkgreen", edgecolor="white")

# Adding a vertical line at the mean score value
plt.axvline(
    scores2.mean(), color="red", linestyle="--", label=f"Mean = {scores2.mean():.3f}"
)

# Set the title and labels for the plot
plt.title("Distribution of Scores for Gilda")
plt.xlabel("Score")
plt.ylabel("Number of entries")
plt.legend()
```
