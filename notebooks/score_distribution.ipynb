```python
# Notebook analyzes the score distribution from the grounding results.
# in `ground_mol_chebi.tsv`.

import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load the grounding results TSV file into a DataFrame and extract the "Score" column.
df = pd.read_csv("../results/ground_mol_chebi.tsv", sep="\t")
df.head()
scores = df["Score"].dropna()  # dropping any NaN values
```

```python
# Plot the score distribution as a histogram using 40 bins
plt.hist(scores, bins=40, color="steelblue", edgecolor="white")

# Adding a vertical line at the mean score value
plt.axvline(
    scores.mean(), color="red", linestyle="--", label=f"Mean = {scores.mean():.3f}"
)

# Set the title and labels for the plot
plt.title("Distribution of Scores")
plt.xlabel("Score")
plt.ylabel("Number of entries")
plt.legend()
```
