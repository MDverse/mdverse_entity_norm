```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load the grounded mollecules
grounded_results = pd.read_csv(
    "../results/ground_molecule/grounded_molecules.tsv", sep="\t"
)
grounded_results.head()
```

```python
# Extracting each entity type percentages
gilda = len(grounded_results[grounded_results["Database"] == "GILDA"]) / len(
    grounded_results
)
pubchem = len(grounded_results[grounded_results["Database"] == "PubChem"]) / len(
    grounded_results
)
chebi = len(grounded_results[grounded_results["Database"] == "CHEBI"]) / len(
    grounded_results
)
pdb = len(grounded_results[grounded_results["Database"] == "PDB"]) / len(
    grounded_results
)
uniprot = len(grounded_results[grounded_results["Database"] == "UNIPROT"]) / len(
    grounded_results
)
seq = len(grounded_results[grounded_results["Database"] == "SEQUENCE"]) / len(
    grounded_results
)

print(
    f"GROUNDING MOLECULE STATISTIQUE : \nPDB: {pdb * 100}%\nUNIPROT:{uniprot * 100}%\n"
    f"CHEBI:{chebi * 100}%\nGILDA:{gilda * 100}%\nPUBCHEM:{pubchem * 100}%\n"
    f"SEQUENCE:{seq * 100}%\n"
)
```

```python
# PLot each percentage
labels = ["PDB", "Uniprot", "Chebi", "Gilda", "PubChem", "Sequence"]
values = [pdb, uniprot, chebi, gilda, pubchem, seq]

fig, ax = plt.subplots(figsize=(6, 4))
bar = ax.bar(labels, values, width=0.50)

ax.set_ylabel("Number of entities")
ax.set_title("Pourcentages of entities per type")
```
